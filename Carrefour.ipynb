### 🔍: Data Cleansing & Integrity Check

في هذه الخطوة، قمنا بمعالجة القيم الشاذة (مثل الأعمار بالسالب والأصفار) وتحويلها إلى قيم فارغة `` لضمان دقة العمليات الإحصائية، مع الحفاظ على كامل حجم الداتا بدون أي فبركة.

________________

```python
df.loc[df['Customer_Age'] <= 0, 'Customer_Age'] = None


df.loc[df['Customer_Rating'] < 1, 'Customer_Rating'] = np.nan
```

```python
# 1. معالجة عمودي Loyalty_Card_Number و Promotion_Name بالقيم النصية المخصصة
df['Loyalty_Card_Number'] = df['Loyalty_Card_Number'].fillna('No_Loyalty_Card')
df['Promotion_Name'] = df['Promotion_Name'].fillna('No_Promotion')
```

```python
# 2. ملء الأعمدة الرقمية المتبقية بالـ Median
numeric_cols = df.select_dtypes(include=['number']).columns
for col in numeric_cols:
    
    df[col] = df[col].fillna(df[col].median())
```

```python
# 3. ملء الأعمدة التصنيفية والنصية المتبقية بالـ Mode (القيمة الأكثر تكراراً)
categorical_cols = df.select_dtypes(include=['O']).columns
for col in categorical_cols:
   
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#4. (معالجة تواريخ الداتا )
df['Date_ &_Time'] = df['Date_ &_Time'].ffill().bfill()
```

```python
# 5. التحقق النهائي من عدم وجود أي مفقودات في الداتا سيت
remaining_nulls = df.isnull().sum().sum()
print(f"إجمالي عدد القيم المفقودة بعد التنظيف = {remaining_nulls}")
```

## 📦 Step 5: Outliers Detection & Handling (فحص ومعالجة القيم الشاذة)
في هذه الخطوة الأخيرة من تنظيف البيانات، سنقوم بما يلي:
1. فحص الأعمدة الرقمية للكشف عن أي قيم شاذة أو متطرفة (`Outliers`) قد تؤثر على دقة التحليل والإحصائيات.
2. استخدام الطرق الإحصائية (مثل IQR - Interquartile Range) أو الرسم لمعرفة مدى انتشار هذه القيم واتخاذ القرار المناسب بشأنها.

```python
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col].count()
    print(f"العمود '{col}': يوجد به حوالي {outliers_count} قيمة شاذة خارج النطاق الطبيعي.")
```

```python
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower_bound,upper=upper_bound)
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
```

## 🛠️ **Step 6: Feature Engineering (هندسة وتجهيز الميزات)**
في هذه المرحلة، سنقوم بإنشاء أعمدة جديدة من البيانات الموجودة لتسهيل عملية التحليل الاستكشافي (EDA) واستخراج رؤى تجارية أقوى:
1. **استخراج خصائص الوقت (`Date & Time Extraction`)**: فصل التاريخ إلى (السنة، الشهر، اليوم، اسم اليوم، وساعة المعاملة).
2. **تجهيز شرائح العملاء (`Customer Age Grouping`)**: تقسيم الأعمار إلى فئات منطقية لتسهيل مقارنة المشتريات حسب الفئة العمرية.

_____________

### 📅 Step 6.1: Extracting Date & Time Features (استخراج تفاصيل التاريخ والوقت)

```python
df['Year'] = df['Date_ &_Time'].dt.year
df['Month'] = df['Date_ &_Time'].dt.month
df['Day'] = df['Date_ &_Time'].dt.day
df['Day_Name'] = df['Date_ &_Time'].dt.day_name()
df['Hour'] = df['Date_ &_Time'].dt.hour
```

### 👥 Step 6.2: Creating Customer Age Groups (تصنيف شرائح أعمار العملاء)

```python
bins = [0, 18, 30, 45, 60, 120]
labels = ['Under 18', '18-30', '31-45', '46-60', '60+']
df['Age_Group'] = pd.cut(df['Customer_Age'], bins=bins, labels=labels)

df['Age_Group'] = df['Age_Group'].astype(str)
```

### 👥 Step 6.3: Customer Rating Transformation (توحيد وتجهيز نسب تقييم العملاء)
تحويل تقييم العملاء من مقياس (من 5) إلى نسبة مئوية (%) لتوحيد نسق المقاييس والمعروضات في البرزنتيشن.

```python
df['Customer_Rating_%'] = (df['Customer_Rating'] / 5) * 100

display(df[['Customer_Rating', 'Customer_Rating_%']].head())
```

```python
# عرض عينة للتأكد من نجاح إضافة الأعمدة الجديدة ومعرفة شكل الداتا
display(df[['Transaction_ID', 'Date_ &_Time', 'Year', 'Month','Day', 'Day_Name', 'Hour', 'Age_Group',"Customer_Rating_%"]].head())
print(f"إجمالي عدد الأعمدة في الداتا بعد التعديل = {df.shape[1]}")
```
