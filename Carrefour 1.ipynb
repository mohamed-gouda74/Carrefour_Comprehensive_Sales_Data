### 🔢 Handling Missing Values in Numerical Columns
Now, we will move on to handling missing values and exploring the **numerical features**.

_________________________

```python
def quick_check(df):
    num = df.select_dtypes('number')
    r = pd.DataFrame({
        'min':        num.min(),
        'median':     num.median(),
        'p99':        num.quantile(.99),
        'max':        num.max(),
        'negatives':  (num < 0).sum(),
        'max_repeats': num.apply(lambda s: (s == s.max()).sum()),
    })
    r['suspect'] = (r['max'] > 3 * r['p99']) | (r['negatives'] > 0) | (r['max_repeats'] > 3)
    return r

quick_check(df)
```

### 🔎: Outliers Detection & Analysis (IQR Method)
In this final inspection step, we detect outliers in our numerical features using the **IQR (Interquartile Range)** method. 
* We calculated $Q1$, $Q3$, and the $IQR$ to establish the normal distribution boundaries ($lower\_bound$ and $upper\_bound$)[cite: 10].
* We quantified how many values fall outside these boundaries for each numerical column[cite: 10].
* Finally, we inspected specific high-value outliers (such as those in `Unit_Price_EGP`) grouped by product categories to determine whether they represent valid premium products or anomalies[cite: 11, 12].

```python
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col].count()
    print(f"العمود '{col}': يوجد به حوالي {outliers_count} قيمة شاذة خارج النطاق الطبيعي.")
```

```python
Q1 = df['Unit_Price_EGP'].quantile(0.25)
Q3 = df['Unit_Price_EGP'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

out = df[df['Unit_Price_EGP'] > upper_bound]
print(len(out))
print(out['Unit_Price_EGP'].describe())
```

## 🛠️ **Step 6: Feature Engineering (هندسة وتجهيز الميزات)**
في هذه المرحلة، سنقوم بإنشاء أعمدة جديدة من البيانات الموجودة لتسهيل عملية التحليل الاستكشافي (EDA) واستخراج رؤى تجارية أقوى:
1. **استخراج خصائص الوقت (`Date & Time Extraction`)**: فصل التاريخ إلى (السنة، الشهر، اليوم، اسم اليوم، وساعة المعاملة).
2. **تجهيز شرائح العملاء (`Customer Age Grouping`)**: تقسيم الأعمار إلى فئات منطقية لتسهيل مقارنة المشتريات حسب الفئة العمرية.

_____________

### 📅 Step 6.1: Extracting Date & Time Features (استخراج تفاصيل التاريخ والوقت)

```python
df['Year'] = df['Date_ &_Time'].dt.year
df['Month'] = df['Date_ &_Time'].dt.month
df['Day'] = df['Date_ &_Time'].dt.day
df['Day_Name'] = df['Date_ &_Time'].dt.day_name()
df['Hour'] = df['Date_ &_Time'].dt.hour
```

### 👥 Step 6.2: Creating Customer Age Groups (تصنيف شرائح أعمار العملاء)

```python
bins = [0, 18, 30, 45, 60, 120]
labels = ['Under 18', '18-30', '31-45', '46-60', '60+']
df['Age_Group'] = pd.cut(df['Customer_Age'], bins=bins, labels=labels)

df['Age_Group'] = df['Age_Group'].astype(str)
```

### 👥 Step 6.3: Customer Rating Transformation (توحيد وتجهيز نسب تقييم العملاء)
تحويل تقييم العملاء من مقياس (من 5) إلى نسبة مئوية (%) لتوحيد نسق المقاييس والمعروضات في البرزنتيشن.

```python
df['Customer_Rating_%'] = (df['Customer_Rating'] / 5) * 100

display(df[['Customer_Rating', 'Customer_Rating_%']].head())
```
